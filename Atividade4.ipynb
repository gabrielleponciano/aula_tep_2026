```python
def selection_sort(vetor):
  for i in range(len(vetor)):
    menor = i

    for j in range(i + 1, len(vetor)):
      if vetor[j] < vetor[menor]:
        menor = j

    vetor[i], vetor[menor] = vetor[menor], vetor[i]

  return vetor

numeros = [5, 2, 8, 1, 3]

resultado = selection_sort(numeros)

print(resultado)
```

```python
def ordenar(vetor, chave):
    for i in range(len(vetor)):
        posicao = i

        for j in range(i + 1, len(vetor)):
            if chave == "crescente":
                if vetor[j] < vetor[posicao]:
                    posicao = j

            elif chave == "decrescente":
                if vetor[j] > vetor[posicao]:
                    posicao = j

        vetor[i], vetor[posicao] = vetor[posicao], vetor[i]

    return vetor

numeros = [5, 2, 8, 1, 3]

print(ordenar(numeros, "crescente"))
print(ordenar(numeros, "decrescente"))
```

```python
def segundo_menor(vetor):
    menor = None
    segundo = None

    for numero in vetor:
        if menor is None or numero < menor:
            segundo = menor
            menor = numero

        elif numero != menor and (segundo is None or numero < segundo):
            segundo = numero

    return segundo


# Exemplo
numeros = [4, 2, 2, 8, 5]

resultado = segundo_menor(numeros)

print("Segundo menor:", resultado)

numeros = [ 6, 3, 8, 5 , 10, 22 ,33 ,33 , 3]

resultado = segundo_menor(numeros)

print("Segundo menor:", resultado)


```

```python
def remover_duplicados(vetor):
    resultado = []

    for numero in vetor:
        if numero not in resultado:
            resultado.append(numero)

    return resultado

numeros = [2, 5, 2, 8, 5, 3, 8]

resultado = remover_duplicados(numeros)

print(resultado)
```

```python

def ordenar_e_contar(vetor):
    # Ordenando em ordem decrescente
    for i in range(len(vetor)):
        maior = i

        for j in range(i + 1, len(vetor)):
            if vetor[j] > vetor[maior]:
                maior = j

        vetor[i], vetor[maior] = vetor[maior], vetor[i]


    pares = 0
    impares = 0

    for numero in vetor:
        if numero % 2 == 0:
            pares += 1
        else:
            impares += 1

    print("Vetor ordenado:", vetor)
    print("Quantidade de pares:", pares)
    print("Quantidade de ímpares:", impares)


numeros = [5, 2, 8, 1, 4, 7, 6]

ordenar_e_contar(numeros)
```

```python

def terceiro_maior(vetor):
    maior = None
    segundo = None
    terceiro = None

    for numero in vetor:
        if numero == maior or numero == segundo or numero == terceiro:
            continue

        if maior is None or numero > maior:
            terceiro = segundo
            segundo = maior
            maior = numero

        elif segundo is None or numero > segundo:
            terceiro = segundo
            segundo = numero

        elif terceiro is None or numero > terceiro:
            terceiro = numero

    return terceiro


# Exemplo
numeros = [10, 8, 8, 5, 3]

resultado = terceiro_maior(numeros)

print("Terceiro maior:", resultado)

```

```python

def mediana(vetor):
    # Ordenando o vetor
    vetor.sort()
    print("Lista ordenada:", vetor) # Nova linha para mostrar o vetor ordenado

    # Encontrando a posição do meio
    meio = len(vetor) // 2

    # Ajuste para vetores com número par de elementos
    if len(vetor) % 2 == 0:
        # Para número par de elementos, a mediana é a média dos dois valores centrais
        return (vetor[meio - 1] + vetor[meio]) / 2
    else:
        # Para número ímpar de elementos, a mediana é o valor central
        return vetor[meio]


# Exemplo
numeros = [7, 2, 9, 4, 5]

resultado = mediana(numeros)

print("Mediana:", resultado)

numeros = [7, 2, 9, 4, 5 , 8]

resultado = mediana(numeros)

print("Mediana:", resultado)
```
